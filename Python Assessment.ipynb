```python
import pandas as pd
import numpy  as np
```

```python
region_a = pd.read_csv("order_region_a.csv")
region_a.head()
```

```python
region_b = pd.read_csv("order_region_b.csv")
region_b.head()
```

```python
region_a.shape, region_b.shape
```

```python
region_a.describe()
```

```python
region_b.describe()
```

```python
region_a['Region'] = 'A'
region_b['Region'] = 'B'
```

```python
total_sale = region_a.append(region_b)
total_sale
```

```python
total_sale[total_sale['OrderId']=='171-0192979-5021168']
```

```python
total_sale = total_sale.drop_duplicates(subset=['OrderId','Region'])
```

```python
total_sale['total_sales'] = total_sale['QuantityOrdered'] * total_sale['ItemPrice']
total_sale
```

```python
total_sale['PromotionDiscount'] = total_sale['PromotionDiscount'].str.extract('([-+]?\d*\.?\d+)').astype(float)
```

```python
total_sale.PromotionDiscount.unique()
```

```python
total_sale['Net_sale'] = total_sale['total_sales'] - total_sale['PromotionDiscount']
```

```python
total_sale = total_sale[total_sale['Net_sale']>0]
```

```python
total_sale[total_sale['Net_sale']<=0].shape
```

```python
total_sale.to_csv('Total_sales.csv', index=False)
```
